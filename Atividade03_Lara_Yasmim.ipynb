```python
numeros = []
pares = []
impares = []

for i in range(10):
    n = int(input("Digite um número:"))
    numeros.append(n)

for n in numeros:
    if n % 2 == 0:
        pares.append(n)
    else:
        impares.append(n)


print("Números:", numeros)
print("Pares:", pares)
print("Ímpares:", impares)
```
