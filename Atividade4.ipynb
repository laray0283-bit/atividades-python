```python
numeros = []
for i in range (10):
    numeros.append(int(input("Número: ")))

atual = maior = 1

for i in range(1, 10):
    if numeros[i] > numeros[i-1]:
        atual +=1
    else:
        atual = 1
    if atual > maior:
        maior = atual

print("Lista:", numeros)
print("Maior sequencia: ", maior)
```

```python
produtos = []

for i in range(5):
    codigo = int(input("Código: "))
    nome = input("Nome: ")
    preco = float(input("Preço: "))
    quantidade = int(input("Quantidade: "))

    produtos.append({"codigo": codigo,"nome":nome,
                      "preco":preco,"quantidade":quantidade})

for i in range(5):
    tipo = input("E ou S: ")
    codigo = int(input("Código: "))
    quantidade = int(input("Quantidade: "))

    for p in produtos:
        if p["codigo"] == codigo and quantidade > 0:
            if tipo == "E":
                p["quantidade"] += quantidade
            elif tipo == "S" and quantidade <= p["quantidade"]:
                p["quantidade"] -= quantidade

total = 0

for p in produtos:
    total += p["preco"]*p["quantidade"]

print("Estoque:", produtos)
print("Total:", total)
```

```python
jogadores = []

for i in range(5):
    nome = input("Nome: ")
    a = int(input("Pontos: "))
    b = int(input("Pontos: "))
    c = int(input("Pontos: "))

    jogadores.append({"nome":nome, "pontos":(a,b,c)})

maior = 0
regular = 999999

for j in jogadores:
    p = j["pontos"]
    j["total"] = p[0]+p[1]+p[2]
    j["media"] = j["total"]/3

    alto = p[0]
    baixo = p[0]

    if p[1] > alto:
        alto = p[1]
    if p[2] > alto:
        alto = p[2]
    if p[1] < baixo:
         baixo = p[1]
    if p[2] < baixo:
         baixo = p[2]

    j["dif"] = alto-baixo

    if j["total"] > maior:
       maior = j["total"]

    if j["dif"] < regular:
        regular = j["dif"]

for j in jogadores:
    print(j["nome"], j["total"],j["media"])

    if j["total"] == maior:
        print("Maior:",j["nome"])

if j["dif"] == regular:
   print("Regular:",j["nome"])
```

```python
votos = {1:0,2:0,3:0,4:0}
lista = []

v = int(input("Voto: "))

while v != 0:
    if v == 1 or v == 2 or v == 3 or v == 4:
        votos[v] += 1
        lista.append(v)
    else:
        print("Inválido")

    v = int(input("Voto: "))

total = votos[1]+votos[2]+votos[3]+votos[4]
maior = votos[1]

for v in votos:
    if votos[v] > maior:
       maior = votos[v]

print("Total:",total)

if total > 0:
   for v in votos:
       print(v,votos[v],votos[v]*100/total,"%")

   for v in votos:
       if votos[v] == maior:
          print("Vencedora:",v)

else:
   print("Nenhum voto")
```

```python
produtos = []

for i in range(5):
    codigo = int(input("Código: "))
    nome = input("Nome: ")
    preco = float(input("Preço: "))
    quantidade = int(input("Quantidade: "))

    produtos.append({"codigo":codigo,"nome":nome,
                     "preco":preco,"quantidade":quantidade,
                     "vendido":0})

historico = []
numero = 0

codigo = int(input("Código: "))

while codigo != 0:
    quantidade = int(input("Quantidade: "))

    for p in produtos:
        if p["codigo"] == codigo and quantidade > 0 and quantidade <= p["quantidade"]:
            p["quantidade"] -= quantidade
            p["vendido"] += quantidade
            historico.append((codigo,quantidade))
            numero += 1

    codigo = int(input("Código: "))

unidades = 0
total = 0

for venda in historico:
    unidades += venda[1]

    for p in produtos:
        if p["codigo"] == venda[0]:
            total += venda[1]*p["preco"]

maior = 0

for p in produtos:
    valor = p["vendido"]*p["preco"]

    if valor > maior:
        maior = valor

print("Estoque:", produtos)
print("Unidades:",unidades)
print("Total:",total)

for p in produtos:
    if p["vendido"]*p["preco"] == maior and maior > 0:
        print("Maior venda:",p["nome"])

print("Histórico:",historico)
print("Resumo:",(numero,unidades,total))
```
