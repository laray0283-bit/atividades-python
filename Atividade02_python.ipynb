```python
secreto = 37
tentativas = 0

while tentativas < 5:
  palpite = int(input("Digite seu palpite: "))
  tentativas += 1

  if palpite == secreto:
      print("Você acertou!")
      break

  if palpite < secreto:
      print("O número secreto é maior")

  if palpite > secreto:
      print("O número secreto é menor")
```

```python
saldo = 200
entrada = 0
retiradas = 0
recusadas = 0
total_entrada = 0
total_saida = 0

codigo = int(input("Digite o código:"))

while codigo != 0:
    valor = float(input("Digite o valor: "))

    if codigo == 1:
        saldo += valor
        entradas += valor
        total_entrada += valor

    elif codigo == 2:
        if valor <= saldo:
           saldo -= valor
           retiradas += valor
           total_saida += valor
        else:
           print("Saldo insuficiente")
           recusadas += 1

    codigo = int(input("Digitr o código:"))

print("Saldo final:", saldo)
print("Quantidade de entradas: ", entradas)
print("Quantidade de retiradas realizadas: ", retiradas)
print("Quantidade de retiradas recusadas:", recusadas)
print("Total que entrou:", total_entrada)
print("Total que saiu:", total_saida)
```
